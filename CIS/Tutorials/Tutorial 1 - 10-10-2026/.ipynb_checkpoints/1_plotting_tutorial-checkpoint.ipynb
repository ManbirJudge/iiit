## 1. Your first x–y plot

To plot a graph we need two things: a list of x values and a list of y values.

````python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)
plt.show()
````

### Always label your graph!

A graph without axis labels is meaningless. Three lines fix that:

````python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.xlabel("x")
plt.ylabel("x^2")
plt.title("Plot")

plt.plot(x, y)
plt.show()
````

### Plotting a mathematical function

Typing every number by hand is boring. `np.linspace(start, stop, n)` gives us `n` equally spaced numbers between `start` and `stop`.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sine Curve")

plt.plot(x, y)
plt.show()
````

> **Try it yourself:** change `100` to `10` in the cell above and run it again. Why does the curve look jagged?

## 2. Several curves and a legend

When there is more than one curve, we need a **legend** to say which is which.

1. Give every curve a `label`
2. Call `plt.legend()`

````python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label="cos y")
plt.plot(x, np.cos(x), label="cos x")

plt.xlabel("x")
plt.title("Sine and Cosine Curve")

plt.legend()
plt.grid(True)
plt.show()
````

### Changing colours, line styles and markers

| What | How | Some options |
|---|---|---|
| Colour | `color="red"` | `"red"`, `"blue"`, `"green"`, `"black"`, `"orange"` |
| Line style | `linestyle="--"` | `"-"` solid, `"--"` dashed, `":"` dotted |
| Marker | `marker="o"` | `"o"` circle, `"s"` square, `"^"` triangle, `"x"` cross |
| Line width | `linewidth=2` | any number |

````python
x = np.linspace(0, 10, 20)

plt.plot(x, np.sin(x), label="cos y", color="red", linestyle="--", marker="o", linewidth=2)
plt.plot(x, np.cos(x), label="cos x", color="black", linestyle=":", marker="s", linewidth=4)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Sine and Cosine Curve")

plt.legend(loc="upper right")
plt.grid(True)
plt.show()
````

### 6.4 Fitting a straight line

Back to the pendulum. Physics tells us

$$T = 2\pi\sqrt{\frac{L}{g}} \quad\Rightarrow\quad T^2 = \frac{4\pi^2}{g}\,L$$

So if we plot $T^2$ against $L$ we should get a **straight line** with slope $4\pi^2/g$.
We can find the best straight line through our points and use its slope to measure $g$!

`np.polyfit(x, y, 1)` finds the best straight line $y = m x + c$ and gives back the slope `m` and the intercept `c`.

## Summary — cheat sheet

| I want to... | Command |
|---|---|
| Line plot | `plt.plot(x, y)` |
| Points only | `plt.scatter(x, y)` |
| Axis labels | `plt.xlabel("...")`, `plt.ylabel("...")` |
| Title | `plt.title("...")` |
| Legend | `plt.plot(x, y, label="...")` then `plt.legend()` |
| Grid | `plt.grid(True)` |
| Save | `plt.savefig("name.png", dpi=300)` |
| Read a data file | `np.loadtxt("file.csv", delimiter=",")` |
| 3D axes | `ax = fig.add_subplot(projection="3d")` |
| Surface | `ax.plot_surface(X, Y, Z)` |
| Histogram | `plt.hist(data, bins=10)` |
| Error bars | `plt.errorbar(x, y, yerr=...)` |
| Mean / std | `np.mean(data)`, `np.std(data)` |
| Straight-line fit | `m, c = np.polyfit(x, y, 1)` |
